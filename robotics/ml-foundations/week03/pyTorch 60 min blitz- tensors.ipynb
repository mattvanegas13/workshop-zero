## Tensors

```python
import torch 
import numpy as np
```

### Tensor initizialization

```python
## From an array or np array
data_x = [[1,2],[3,4]]
tensor_x = torch.tensor(data_x)
numpy_data_x = np.array(data_x)
tensor_from_numpy = torch.tensor(numpy_data_x)
print(tensor_from_numpy)
print(tensor_x)
```

```python
# From another tensor

## Notice that we retain the shape and data type but not the data if we use ones_like
## We could override the the data type as well if we'd  like
print(torch.tensor(tensor_x))
print(torch.ones_like(tensor_x))
print(torch.rand_like(tensor_x, dtype=torch.float))
```

```python
# what is shape?
# shape determines the dimensionality of the vector
shape = (2,3,)
torch.ones(shape)
```

### Attributes

```python
print(f"Shape of tensor: {tensor_x.shape}")
print(f"Datatype of tensor: {tensor_x.dtype}")
print(f"Device tensor is stored on: {tensor_x.device}") # where on memory tensor is stored
```

### Operations 

```python
## Similar to numpy we can multiply tensors in the following way ( element wise product )
t1 = torch.rand((2,3,))
t2 = torch.rand((2,3,))
t3 = torch.rand((3,2,))
print(t1)
print()
print(t2)
print()
print(t3)
print()
print(t1*t2)
print()
## Matrix multiplication 
print(t1@t3)

## Indexing works the following way

##  We can also concat them along various index
```

```python
## Indexing works the following way
print(t1[0])
print(t1[1])
print(t1[0,2])
print(t1[0:,1])
print(t1[1:,1])
```

```python
# Joining tensors
print(torch.cat([tensor_x,tensor_x], dim=1))
```

## Important Caveats

### Using _ suffixed operations

```python
# In-place operations
# there are some operations that have a _ suffix -> these change the tensor its called on
print(tensor_x)
y = tensor_x.add_(5)
```

### Modifying tensor and numpy

```python
# Tensor and numpy arrays spawned from them share the same memory location - so
# modification of one results in the modification of the other
```

```python
t = torch.ones(5)
n = t.numpy()
print(t)
print(n)
```

```python
t.add_(3)
print(n)
```

```python
# similarly changes in the numpy array affect the tensor
np.add(n,1, out=n)
print(t)
```

```python
n += 2
print(t)
```
