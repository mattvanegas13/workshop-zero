## Tensors

```python
import torch 
import numpy as np
```

### Tensor initizialization

```python
## From an array or np array
data_x = [[1,2],[3,4]]
tensor_x = torch.tensor(data_x)
numpy_data_x = np.array(data_x)
tensor_from_numpy = torch.tensor(numpy_data_x)
print(tensor_from_numpy)
print(tensor_x)
```

```python
# From another tensor
tensor_from_tensor = torch.ones_like(tensor_x)
tensor_from_tensor
## Notice that we retain the shape and data type but not the data
## We could override the the data type as well if we'd  like
torch.rand_like(tensor_x, dtype=torch.float)
```
