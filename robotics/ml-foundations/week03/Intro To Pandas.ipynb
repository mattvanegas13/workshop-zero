```python
import pandas as pd
```

```python
## Series Construction -> dict and array baby
```

```python
## construction from dict
ret_dict = {"AAPL": -0.01}
pd.Series(ret_dict)
```

```python
## from two lists
ret = [-0.01, 2]
tickers = ["AAPL","MSFT"]
s = pd.Series(ret,index=tickers)
```

```python
s.keys()
```

```python
# retrievals
# from key
s['AAPL']
```

```python
# from several keys
s[["AAPL", "MSFT"]]
```

```python
s[s>0]
```

```python
# Modifying values
s[['AAPL', 'MSFT']] = [0,0]
```

```python
s
```

```python
# Series Methods
ret_dict = {'aapl':-.1,'msft':-.02,'tsla':0.015}
s = pd.Series(ret_dict)
s
```

```python
s - .05
```

```python
s.map(lambda x: x if x>.5 else x**2)
```

```python
s.rank()
```

```python
## data alignment
y = pd.Series({'aapl':.2, 'lulu':.3})
s + y
```

```python
s.add(y, fill_value=0)
```
